```python
from PIL import Image
import numpy as np

img = Image.open("company-logo-header.png").convert("RGBA")
data = np.array(img)

r, g, b, a = data[:,:,0], data[:,:,1], data[:,:,2], data[:,:,3]

# Xóa nền: pixel nào gần trắng (>200,>200,>200) hoặc gần xám ca rô (~190,190,190)
# thì set alpha = 0
is_white = (r > 200) & (g > 200) & (b > 200)
is_checker_light = (r > 180) & (g > 180) & (b > 180)
is_checker_dark  = (r > 140) & (g > 140) & (b > 140) & (r < 200) & (g < 200) & (b < 200)

mask = is_white | is_checker_light | is_checker_dark
data[:,:,3] = np.where(mask, 0, a)

result = Image.fromarray(data)
result.save("company-logo-header.png", "PNG")
print("Done. Image size:", result.size)
```
